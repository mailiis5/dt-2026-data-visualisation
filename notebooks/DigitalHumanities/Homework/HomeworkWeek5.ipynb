```python
import pandas as pd
df = pd.read_csv("Bookdatabase.csv")
df.head()
#df.describe() 
```

```python
import matplotlib.pyplot as plt 
df.plot(kind="bar", x="Stephen_King_book", y="Pages", legend=False, color="#CBC3E3")
#plt.show() 
plt.title("Lenght of Stephen King books")
plt.ylabel("Pages")
```

On this chart we can see the lenght of Stephen King's books. Some of the longer books bein "It" and "Under the Dome" and some of the shorter ones being "Carrie" and #The Gunslinger". We can also notice that most of his books are around 400-600 pages.

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Stephen_King_book", y="Rating", legend=False, color="#CBC3E3")
#plt.show() 
plt.title("Rating per book")
plt.ylabel("Rating")
```

This barchart shows the ratings of the books (based on GoodReads). The ratings are quite stable and collectively close to a 4. The best rated is "The Green Mile" and the worst rated is "Cell".

```python
pages_by_month = df.groupby("Publishing_Month")["Pages"].sum()

pages_by_month.plot(kind="pie",labels=pages_by_month.index, legend=False, autopct="%1.1f%%")

plt.title("Pages Written by Month")
plt.ylabel("")  # removes the default y-axis label
plt.show()
```

This chart shows which months has Stephen King published his books. The most popular publishing month is September with 28.6% of books being published then. Note that no books have been published in the months of February, March, July and December.

```python
plt.scatter(df["Decade"], df["Rating"], s=df["Pages"]*0.5, alpha=0.4)

plt.title("Book Ratings and Lenghts over the decades")
plt.ylabel("Rating")
plt.xlabel("Decade")
```

Here we can see how well were the books from different decades rated and also the lenght of the books (by the size of the bubbles). The 2000s and 2010s has the biggest differeces of book ratings. The 2020s (so far) is the most stable with the ratings.

```python
avg_pages= df.groupby("Decade")["Pages"].mean()
avg_pages.plot(kind="bar")
plt.show()
```

On this chart we can see how many pages, on average, is a book from a certain decade. From this we can conclude that the books written in the 1970s tend to be a bit shorter than most, and the books from the 1990s tend to be a bit longer.

```python
books_by_decade = df.groupby("Decade").size()

books_by_decade.plot(kind="bar", color="#CBC3E3")

plt.title("Books Published by Decade")
plt.xlabel("Decade")
plt.ylabel("Number of Books")

plt.show()
```

Here we can see how many books were published per decade. The most books were published in the 1980s and the least in the 1990s and 2020s (although the 2020s statistic might change since it is still ongoing).
