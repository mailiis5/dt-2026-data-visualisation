create a chart showing who puts in the most effort. Effort can be find from number of hours.

```python
import pandas as pd
df = pd.read_csv("Competencies.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
#colors= pt.cm.plasma (range(len(avg_hours)))
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", color="#CBC3E3")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Grade"]*5, alpha=0.4, color="#CBC3E3")
plt.title("Hours studied per grade")
plt.ylabel("Grade")
plt.xlabel("Hours studied")
```

```python
plt.Figure(figsize=(8,5))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot for hours studied and grades")
plt.show
```
